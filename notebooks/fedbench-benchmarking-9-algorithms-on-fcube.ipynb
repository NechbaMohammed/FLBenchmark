```python
def extract_data(history):
    centralized_metrics = history.metrics_centralized
    rounds = [round_num for round_num, _ in centralized_metrics["test_accuracy"]]
    accuracies = [value for _, value in centralized_metrics["test_accuracy"]]
    return rounds, accuracies
```

```python
def plot_federated_learning_curves(
    fedavg_history,
    fedprox_history,
    fedadam_history,
    fedadagrad_history,
    fedyogi_history,
    fednova_history,
    fedscaffold_history,
    fedmoon_history,
    fedbn_history,
    save_plot=False,
    filename='federated_learning_plot.png'
):
    # Extract data
    fedavg_rounds, fedavg_accuracies = extract_data(fedavg_history)
    fedprox_rounds, fedprox_accuracies = extract_data(fedprox_history)
    fedadam_rounds, fedadam_accuracies = extract_data(fedadam_history)
    fedadagrad_rounds, fedadagrad_accuracies = extract_data(fedadagrad_history)
    fedyogi_rounds, fedyogi_accuracies = extract_data(fedyogi_history)
    fednova_rounds, fednova_accuracies = extract_data(fednova_history)
    fedscaffold_rounds, fedscaffold_accuracies = extract_data(fedscaffold_history)
    fedmoon_rounds, fedmoon_accuracies = extract_data(fedmoon_history)
    fedbn_rounds, fedbn_accuracies = extract_data(fedbn_history)

    # Plotting
    plt.figure(figsize=(8, 6), dpi=150)

    plt.plot(fedavg_rounds, fedavg_accuracies, label='FedAvg', color='#1f77b4', linestyle='-',marker='^', markersize=6)
    plt.plot(fedprox_rounds, fedprox_accuracies, label='FedProx, μ=0.1', color='#ff7f0e', linestyle='--',marker='*',markersize=6)
    plt.plot(fedadam_rounds, fedadam_accuracies, label='Adam', color='#2ca02c', linestyle=':',marker='o',markersize=6)
    plt.plot(fedadagrad_rounds, fedadagrad_accuracies, label='FedAdagrad', color='#d62728', linestyle='-.',marker='s',markersize=6)
    plt.plot(fedyogi_rounds, fedyogi_accuracies, label='FedYogi', color='#9467bd', linestyle=(0, (3, 1, 1, 1)),marker='D',markersize=6)
    plt.plot(fednova_rounds, fednova_accuracies, label='FedNova', color='#8c564b', linestyle=(0, (5, 1)),marker='P',markersize=6)
    plt.plot(fedscaffold_rounds, fedscaffold_accuracies, label='FedScaffold', color='#e377c2', linestyle=(0, (1, 1)),marker='X',markersize=6)
    plt.plot(fedmoon_rounds, fedmoon_accuracies, label='FedMoon', color='#7f7f7f', linestyle=(0, (3, 5, 1, 5)),marker='H',markersize=6)
    plt.plot(fedbn_rounds, fedbn_accuracies, label='FedBN', color='#17becf', linestyle= (0, (5, 5)),marker='>',markersize=6)
    # Labels and title
    plt.xlabel('Communication round', fontsize=20)
    plt.ylabel('Test acc', fontsize=20)
    plt.xticks(fontsize=18)
    plt.yticks(fontsize=18)

    # Legend
    plt.legend(loc='lower right', fontsize=18)

    plt.tight_layout()

    # Save plot (optional)
    if save_plot:
        plt.savefig(filename)  # Saves the plot to a file

    # Show plot
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()
```

```python
plot_federated_learning_curves(
    fedavg_history,
    fedprox_history,
    fedadam_history,
    fedadagrad_history,
    fedyogi_history,
    fednova_history,
    fedscaffold_history,
    fedmoon_history,
    fedbn_history,
    save_plot=True,
    filename='Feature_distribution_skew.png'
)
```
